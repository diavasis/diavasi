Java client `dev.diavasi:diavasi-data:0.1.0`. Put that artifact on the IJava classpath, then call `DiavasiClient.consume`.


```java
import dev.diavasi.data.DiavasiClient;
var options = new DiavasiClient.Options();
options.addr = System.getenv("DIAVASI_DATA_ADDR");
options.ca = System.getenv("DIAVASI_CA");
options.token = System.getenv("DIAVASI_API_TOKEN");
options.groupId = "demo";
options.consumerId = "notebook-java";
options.expectRecords = 8;
DiavasiClient.consume(options).batchIds
```
