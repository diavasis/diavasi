The Go client module is mounted at `/opt/diavasi/clients/go`.


```go
import (
    "context"
    "fmt"
    "os"
    "github.com/diavasis/diavasi/clients/go"
)
report, err := diavasi.Consume(context.Background(), diavasi.Options{
    Addr: os.Getenv("DIAVASI_DATA_ADDR"),
    CAFile: os.Getenv("DIAVASI_CA"),
    Token: os.Getenv("DIAVASI_API_TOKEN"),
    GroupID: "demo",
    ConsumerID: "notebook-go",
    ExpectRecords: 8,
})
fmt.Println(report.BatchIDs, err)
```
