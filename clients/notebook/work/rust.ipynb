Load `diavasi-client` from the mounted tree and consume the synthetic group.


```rust
:dep diavasi-client = { path = "/opt/diavasi/clients/rust" }
let pem = std::fs::read(std::env::var("DIAVASI_CA").unwrap()).unwrap();
let mut opts = diavasi_client::Options::new(
    std::env::var("DIAVASI_DATA_ADDR").unwrap(),
    pem,
    std::env::var("DIAVASI_API_TOKEN").unwrap(),
    "demo",
    "notebook-rust",
);
opts.expect_records = Some(8);
let report = diavasi_client::run(opts).unwrap();
report.batch_ids
```
